## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

For product editors, deciding which recommendations to serve, we will build a classification model from user interaction logs, predicting ⁠is_engaged⁠ measured by ROC-AUC and F1-Score. A wrong call costs wasted recommendation slots and lower user retention. A plain rule isn't enough because user behavior patterns are too complex and shift over time.

A plain heuristic rule (e.g., "If user activity > 10, recommend item") fails because real user behavior patterns are tangled, noisy, non-linear, and shift over time. Fixed rules cannot personalize at scale or adapt dynamically. Machine Learning earns its place by continuously learning complex interactions across multiple signals that are too messy to write by hand.
